```python
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.metrics.pairwise import cosine_similarity

from gensim.models import Word2Vec
from gensim.models import FastText
```

```python
documents = [
    "I love Python programming",
    "Python is easy to learn",
    "I love learning NLP",
    "NLP is interesting and useful"
]

print(documents)
```

```python
for i, document in enumerate(documents):
    print("Document", i + 1, ":", document)
```

```python
enumerate(documents)
```

```python
vectorizer = CountVectorizer()
vectorizer
```

```python
bow_matrix = vectorizer.fit_transform(documents)
bow_matrix
```

```python
vocabulary = vectorizer.get_feature_names_out()

print(vocabulary)
```

```python
print(bow_matrix.toarray())
```

```python
bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vocabulary
)

bow_df
```

```python
binary_vectorizer = CountVectorizer(binary=True)

binary_matrix = binary_vectorizer.fit_transform(documents)

print(binary_matrix.toarray())
```

```python
unigram_vectorizer = CountVectorizer(
    ngram_range=(1, 1)
)

unigram_vectorizer.fit(documents)

print(unigram_vectorizer.get_feature_names_out())
```

```python
bigram_vectorizer = CountVectorizer(
    ngram_range=(2, 2)
)

bigram_vectorizer.fit(documents)

print(bigram_vectorizer.get_feature_names_out())
```

```python
trigram_vectorizer = CountVectorizer(
    ngram_range=(3, 3)
)

trigram_vectorizer.fit(documents)

print(trigram_vectorizer.get_feature_names_out())
```

```python
tfidf_vectorizer = TfidfVectorizer()
```

```python
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)
```

```python
tfidf_vocabulary = tfidf_vectorizer.get_feature_names_out()

print(tfidf_vocabulary)
```

```python
print(tfidf_matrix.toarray())
```

```python
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vocabulary
)

tfidf_df.round(3)
```

```python
sentences = [
    ["i", "love", "python", "programming"],
    ["python", "is", "easy", "to", "learn"],
    ["i", "love", "learning", "nlp"],
    ["nlp", "is", "interesting", "and", "useful"],
    ["python", "is", "powerful"],
    ["nlp", "is", "useful"],
    ["i", "enjoy", "python"],
    ["i", "enjoy", "nlp"]
]

for sentence in sentences:
    print(sentence)
```

```python
cbow_model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=3,
    min_count=1,
    sg=0,
    workers=1,
    seed=42
)
```

```python
python_vector = cbow_model.wv["python"]

print(python_vector)
```

```python
print(python_vector.shape)
```

```python
similar_words = cbow_model.wv.most_similar(
    "python",
    topn=5
)

print(similar_words)
```

```python
skipgram_model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=3,
    min_count=1,
    sg=1,
    workers=1,
    seed=42
)
```

```python
nlp_vector = skipgram_model.wv["nlp"]

print(nlp_vector)
```

```python
print(nlp_vector.shape)
```

```python
similarity = skipgram_model.wv.similarity(
    "python",
    "programming"
)

print(similarity)
```

```python
fasttext_model = FastText(
    sentences=sentences,
    vector_size=100,
    window=3,
    min_count=1,
    workers=1,
    seed=42
)
```

```python
fasttext_vector = fasttext_model.wv["python"]

print(fasttext_vector)
```

```python
print(fasttext_vector.shape)
```

```python
fasttext_similar = fasttext_model.wv.most_similar(
    "python",
    topn=5
)

print(fasttext_similar)
```

```python
new_word = "pythonprogramming"

vector = fasttext_model.wv[new_word]

print(vector)
```

```python
python_vector = skipgram_model.wv["python"]
nlp_vector = skipgram_model.wv["nlp"]

similarity = cosine_similarity(
    [python_vector],
    [nlp_vector]
)

print(similarity)
```

```python
sentence1 = ["I love Python programming"]

sentence2 = ["Python programming is interesting"]

combined_sentences = sentence1 + sentence2

vectorizer = TfidfVectorizer()

vectors = vectorizer.fit_transform(combined_sentences)

print(vectors.toarray())
```

```python
similarity = cosine_similarity(vectors)

print(similarity)
```
