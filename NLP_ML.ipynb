```python
# Naive Bayes
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

documents = [
    "I love this movie",
    "This movie is excellent",
    "I hate this movie",
    "This movie is terrible"
]

labels = [
    "positive",
    "positive",
    "negative",
    "negative"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

model = MultinomialNB()

model.fit(X, labels)

new_text = ["I love this excellent movie"]

new_X = vectorizer.transform(new_text)

prediction = model.predict(new_X)

print(prediction)
```

```python
#logistic_regression
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(X, labels)

prediction = model.predict(new_X)

print(prediction)
```

```python
#SVM
```

```python
from sklearn.svm import LinearSVC

model = LinearSVC()

model.fit(X, labels)

prediction = model.predict(new_X)

print(prediction)
```

```python
#Random forest
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X, labels)

prediction = model.predict(new_X)

print(prediction)
```

```python
#K-Means
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

documents = [
    "football match player goal",
    "cricket match team score",
    "stock market shares finance",
    "company profit business",
    "machine learning artificial intelligence",
    "AI model deep learning"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X)

print(clusters)
```
