```python
import spacy

# Load English NLP model
nlp = spacy.load("en_core_web_sm")

# Input text
text = """
Apple CEO Tim Cook visited Chennai on January 15, 2026.
He discussed Apple's new products with Microsoft executives.
"""

# Process text
doc = nlp(text)

# Extract named entities
for ent in doc.ents:

    print(
        "Entity:",
        ent.text,
        "| Label:",
        ent.label_,
        "| Meaning:",
        spacy.explain(ent.label_)
    )
```

```python
import spacy
import pandas as pd
from collections import Counter
from spacy import displacy


# ============================================================
# 1. LOAD SPACY MODEL
# ============================================================

nlp = spacy.load("en_core_web_sm")


# ============================================================
# 2. INPUT TEXT
# ============================================================

text = """
Jaison Immanuel is a data analyst from Chennai, India.
He completed his MSc in Business Statistics at VIT University in 2026.
He has experience working with Python, SQL, Power BI, Tableau,
Pandas, NumPy, Scikit-learn and spaCy.

In September 2026, Jaison joined GeakMinds as an Associate Python Developer.
The company works with technologies such as AWS, Azure and Google Cloud.
His manager assigned him a cloud cost optimization project.

Jaison attended a conference in Bengaluru on January 15, 2026.
The conference was organized by Microsoft and Amazon.
The project budget was $50,000 and the expected improvement was 25 percent.

Later, he travelled from Chennai to New Delhi for a meeting.
The meeting started at 10:30 AM on March 20, 2026.
He purchased an iPhone for 80,000 Indian rupees.
"""


# ============================================================
# 3. PROCESS THE COMPLETE TEXT
# ============================================================

doc = nlp(text)


# ============================================================
# 4. DISPLAY ORIGINAL TEXT
# ============================================================

print("=" * 80)
print("ORIGINAL TEXT")
print("=" * 80)

print(text)


# ============================================================
# 5. DISPLAY ALL TOKENS
# ============================================================

print("\n")
print("=" * 80)
print("ALL TOKENS")
print("=" * 80)

for token in doc:

    print(
        f"Token: {token.text:<20} "
        f"POS: {token.pos_:<10} "
        f"Lemma: {token.lemma_:<20} "
        f"Stopword: {token.is_stop}"
    )


# ============================================================
# 6. DISPLAY SENTENCES
# ============================================================

print("\n")
print("=" * 80)
print("SENTENCES")
print("=" * 80)

for number, sentence in enumerate(doc.sents, start=1):

    print(f"\nSentence {number}:")
    print(sentence.text)


# ============================================================
# 7. EXTRACT ALL NAMED ENTITIES
# ============================================================

print("\n")
print("=" * 80)
print("NAMED ENTITIES")
print("=" * 80)

for ent in doc.ents:

    print(
        f"Entity: {ent.text:<30} "
        f"Label: {ent.label_:<15} "
        f"Description: {spacy.explain(ent.label_)}"
    )


# ============================================================
# 8. ENTITY DETAILS
# ============================================================

print("\n")
print("=" * 80)
print("DETAILED ENTITY INFORMATION")
print("=" * 80)

for number, ent in enumerate(doc.ents, start=1):

    print(f"\nEntity {number}")

    print("Text       :", ent.text)
    print("Label      :", ent.label_)
    print("Meaning    :", spacy.explain(ent.label_))
    print("Start token:", ent.start)
    print("End token  :", ent.end)


# ============================================================
# 9. CREATE ENTITY DATAFRAME
# ============================================================

entity_data = []

for ent in doc.ents:

    entity_data.append({
        "Entity": ent.text,
        "Label": ent.label_,
        "Meaning": spacy.explain(ent.label_),
        "Start": ent.start_char,
        "End": ent.end_char
    })


entity_df = pd.DataFrame(entity_data)


print("\n")
print("=" * 80)
print("ENTITY DATAFRAME")
print("=" * 80)

print(entity_df.to_string(index=False))


# ============================================================
# 10. COUNT ENTITY TYPES
# ============================================================

print("\n")
print("=" * 80)
print("ENTITY TYPE COUNTS")
print("=" * 80)

entity_counts = Counter(ent.label_ for ent in doc.ents)

for label, count in entity_counts.items():

    print(
        f"{label:<15} "
        f"{count:<5} "
        f"{spacy.explain(label)}"
    )


# ============================================================
# 11. DISPLAY ENTITIES GROUPED BY TYPE
# ============================================================

print("\n")
print("=" * 80)
print("ENTITIES GROUPED BY TYPE")
print("=" * 80)

grouped_entities = {}

for ent in doc.ents:

    if ent.label_ not in grouped_entities:

        grouped_entities[ent.label_] = []

    grouped_entities[ent.label_].append(ent.text)


for label, entities in grouped_entities.items():

    print(f"\n{label} - {spacy.explain(label)}")

    for entity in entities:

        print("   ->", entity)


# ============================================================
# 12. REMOVE DUPLICATE ENTITIES
# ============================================================

print("\n")
print("=" * 80)
print("UNIQUE ENTITIES")
print("=" * 80)

unique_entities = {}

for ent in doc.ents:

    key = (ent.text, ent.label_)

    unique_entities[key] = True


for entity, label in unique_entities:

    print(
        f"{entity:<30} -> {label}"
    )


# ============================================================
# 13. ENTITY RECOGNITION SENTENCE BY SENTENCE
# ============================================================

print("\n")
print("=" * 80)
print("ENTITY RECOGNITION SENTENCE BY SENTENCE")
print("=" * 80)

for number, sentence in enumerate(doc.sents, start=1):

    print(f"\nSentence {number}:")
    print(sentence.text)

    if sentence.ents:

        print("Entities:")

        for ent in sentence.ents:

            print(
                f"   {ent.text} -> "
                f"{ent.label_} "
                f"({spacy.explain(ent.label_)})"
            )

    else:

        print("No named entities found.")


# ============================================================
# 14. CHECK EVERY TOKEN FOR ENTITY INFORMATION
# ============================================================

print("\n")
print("=" * 80)
print("TOKEN-LEVEL ENTITY INFORMATION")
print("=" * 80)

for token in doc:

    if token.ent_type_:

        print(
            f"Token: {token.text:<20} "
            f"Entity Type: {token.ent_type_:<15} "
            f"Entity IOB: {token.ent_iob_}"
        )


# ============================================================
# 15. BIO / IOB STYLE INFORMATION
# ============================================================

print("\n")
print("=" * 80)
print("BIO / IOB ENTITY TAGGING")
print("=" * 80)

for token in doc:

    print(
        f"{token.text:<25} "
        f"{token.ent_iob_:<5} "
        f"{token.ent_type_}"
    )


# ============================================================
# 16. FIND PERSON ENTITIES
# ============================================================

print("\n")
print("=" * 80)
print("PERSON ENTITIES")
print("=" * 80)

for ent in doc.ents:

    if ent.label_ == "PERSON":

        print("PERSON:", ent.text)


# ============================================================
# 17. FIND ORGANIZATIONS
# ============================================================

print("\n")
print("=" * 80)
print("ORGANIZATION ENTITIES")
print("=" * 80)

for ent in doc.ents:

    if ent.label_ == "ORG":

        print("ORGANIZATION:", ent.text)


# ============================================================
# 18. FIND LOCATIONS / PLACES
# ============================================================

print("\n")
print("=" * 80)
print("LOCATION / GPE ENTITIES")
print("=" * 80)

for ent in doc.ents:

    if ent.label_ in ["GPE", "LOC", "FAC"]:

        print(
            f"{ent.text} -> {ent.label_}"
        )


# ============================================================
# 19. FIND DATE ENTITIES
# ============================================================

print("\n")
print("=" * 80)
print("DATE ENTITIES")
print("=" * 80)

for ent in doc.ents:

    if ent.label_ == "DATE":

        print("DATE:", ent.text)


# ============================================================
# 20. FIND MONEY ENTITIES
# ============================================================

print("\n")
print("=" * 80)
print("MONEY ENTITIES")
print("=" * 80)

for ent in doc.ents:

    if ent.label_ == "MONEY":

        print("MONEY:", ent.text)


# ============================================================
# 21. FIND PERCENTAGE ENTITIES
# ============================================================

print("\n")
print("=" * 80)
print("PERCENTAGE ENTITIES")
print("=" * 80)

for ent in doc.ents:

    if ent.label_ == "PERCENT":

        print("PERCENT:", ent.text)


# ============================================================
# 22. NER SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("NER SUMMARY")
print("=" * 80)

print("Total characters :", len(text))
print("Total tokens     :", len(doc))
print("Total sentences  :", len(list(doc.sents)))
print("Total entities   :", len(doc.ents))
print("Unique entities  :", len(unique_entities))


# ============================================================
# 23. SAVE ENTITY INFORMATION TO CSV
# ============================================================

if not entity_df.empty:

    entity_df.to_csv(
        "ner_entities.csv",
        index=False
    )

    print("\nEntity information saved to:")
    print("ner_entities.csv")


# ============================================================
# 24. VISUALIZE NER
# ============================================================

print("\n")
print("=" * 80)
print("OPENING NER VISUALIZATION")
print("=" * 80)

displacy.serve(
    doc,
    style="ent",
    port=5000
)
```
